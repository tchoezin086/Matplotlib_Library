```python

import matplotlib.pyplot as plt
months=["jan","feb","mar","april","may","june"]
sales=[100,200,300,400,500,600]
plt.plot(months,sales)
plt.xlabel("months")
plt.ylabel("sales")
plt.title("monthly sales")
plt.show()
```

```python
import matplotlib.pyplot as plt
year = [2021, 2022, 2023, 2024, 2025, 2026]
sales = [20000, 30000, 25000, 40000, 50000, 65000]
plt.plot(year,sales)
plt.xlabel("year")
plt.ylabel("sales")
plt.title("Yearly Sales Growth")
plt.show()
```

```python
City=["Delhi","Pune","Maumbai"]

Sales=[1000,5000,2000]

plt.bar(City,Sales)

plt.xlabel("City")

plt.ylabel("Sales")

plt.title("City vs  sales")

plt.show()
```

```python
import matplotlib.pyplot as plt
employees = ["Amit", "Priya", "Rahul", "Neha", "Vikas"]
salary = [30000, 40000, 45000, 55000, 35000]
plt.bar(employees,salary)
plt.xlabel("employees")
plt.ylabel("salary")
plt.title("Employee salary")
plt.show()
```

```python
import matplotlib.pyplot as plt
courses = ["Python", "SQL", "Java", "Excel", "Power BI"]
students = [40, 30, 25, 35, 20]
plt.bar(courses,students)
plt.xlabel("courses")
plt.ylabel("students")
plt.title("Students Enrolled in Different Courses")
plt.show()
```

```python
import matplotlib.pyplot as plt
expenses=[5000,1000,2000,5000,6000]
Type=["Food","Travel","Rent","Shopping","Other"]
plt.pie(expenses,labels=Type,autopct="%1.1f%%")
plt.title("Expenses")
plt.show()

```

```python
import matplotlib.pyplot as plt
activities = ["Sleeping","Studying","College","Entertainment","Eating","Other"]
hours = [8, 4, 6, 2, 2, 2]
plt.pie(hours,labels=activities,autopct="%1.1f%%")
plt.title("Daily Activities")
plt.show()
```

```python
Adv=[100,200,300,400,500]
Sales=[1000,2000,3000,4000,5000]
plt.scatter(Adv,Sales)
plt.xlabel("Advertisement")
plt.ylabel("Sales")
plt.title("Advertisement vs Sales")
plt.show()
```

```python
#Negative correlation

Adv=[100,200,300,400,500]
Sales=[5000,4000,3000,2000,1000]
plt.scatter(Adv,Sales)
plt.plot(Adv,Sales)  # to show line
plt.xlabel("Advertisement")
plt.ylabel("Sales")
plt.title("Advertisement vs Sales")
plt.show()
```

```python
#My example:
product_price = [200, 500, 800, 1000, 1500, 2000, 2500, 3000, 3500, 4000]
customer_rating = [3.5, 4.0, 4.2, 3.8, 4.5, 4.1, 4.7, 4.3, 4.8, 4.4]
plt.scatter(product_price,customer_rating)
plt.plot(product_price,customer_rating)  # to show line
plt.xlabel("Price")
plt.ylabel("Rating")
plt.title("Product price vs Customer rating")
plt.show()
```
